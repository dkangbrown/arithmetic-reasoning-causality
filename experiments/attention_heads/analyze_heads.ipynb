```python
import pandas as pd

prob_csv = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/logit_lens_h_pre_penultimate_sum_-1_2.csv"
sim_csv  = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/residual_similarity_h_pre_penultimate_sum_-1.csv"

prob_df = pd.read_csv(prob_csv)
sim_df  = pd.read_csv(sim_csv)


def top_64_head_set(df, col):
    """Return the set of top-64 unique (layer, head_num) pairs for a given column."""
    sorted_df = (
        df[['layer', 'head_num', col]]
        .sort_values(col, ascending=False)
        .drop_duplicates(subset=['layer', 'head_num'])
        .head(64)
    )
    return {(int(r['layer']), int(r['head_num'])) for _, r in sorted_df.iterrows()}


bases = ['operand_1_1', 'operand_1_2', 'operand_2', 'factual_output']
metric_sources = [
    ('prob',    prob_df),
    ('cos_sim', sim_df),
    ('proj',    sim_df),
]

# Collect top-64 head sets for every (base, metric) combination
heads_by_metric = {}
for base in bases:
    for suffix, df in metric_sources:
        col = f"{base}_{suffix}"
        heads_by_metric[(base, suffix)] = top_64_head_set(df, col)

# Pairwise comparisons requested:
#   *_prob    vs *_cos_sim
#   *_cos_sim vs *_proj
#   *_prob    vs *_proj
pairs = [('prob', 'cos_sim'), ('cos_sim', 'proj'), ('prob', 'proj')]

for base in bases:
    print(f"=== {base} ===")
    for s1, s2 in pairs:
        set1 = heads_by_metric[(base, s1)]
        set2 = heads_by_metric[(base, s2)]
        shared = set1 & set2
        print(f"  {base}_{s1} vs {base}_{s2}: {len(shared)}/64 shared heads")
        # for layer, head in sorted(shared):
        #     print(f"    (Layer: {layer}, Head: {head})")
    print()
```

```python
import pandas as pd

prob_csv = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/logit_lens_h_pre_penultimate_sum_-1.csv"
intervened_prob_csv = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/intervened_logit_lens_h_pre_penultimate_sum_-1.csv"
sim_csv  = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/residual_similarity_h_pre_penultimate_sum_-1.csv"
intervened_sim_csv = f"{project_root}/experiments/attention_heads/output/GPT-OSS_stepwise/head_activation/intervened_residual_similarity_h_pre_penultimate_sum_-1.csv"

prob_df = pd.read_csv(prob_csv)
intervened_prob_df = pd.read_csv(intervened_prob_csv)
sim_df  = pd.read_csv(sim_csv)
intervened_sim_df  = pd.read_csv(intervened_sim_csv)


def top_64_head_set(df, col):
    """Return the set of top-64 unique (layer, head_num) pairs for a given column."""
    sorted_df = (
        df[['layer', 'head_num', col]]
        .sort_values(col, ascending=False)
        .drop_duplicates(subset=['layer', 'head_num'])
        .head(64)
    )
    return {(int(r['layer']), int(r['head_num'])) for _, r in sorted_df.iterrows()}


bases = ['operand_1_1', 'operand_1_2', 'operand_2', 'factual_output']
# (suffix, regular_df, intervened_df)
metric_sources = [
    ('prob',    prob_df, intervened_prob_df),
    ('cos_sim', sim_df,  intervened_sim_df),
    ('proj',    sim_df,  intervened_sim_df),
]

for base in bases:
    print(f"=== {base} ===")
    for suffix, reg_df, int_df in metric_sources:
        col = f"{base}_{suffix}"
        reg_set = top_64_head_set(reg_df, col)
        int_set = top_64_head_set(int_df, col)
        shared = reg_set & int_set
        print(f"  {col} (regular vs intervened): {len(shared)}/64 shared heads")
        # for layer, head in sorted(shared):
        #     print(f"    (Layer: {layer}, Head: {head})")
    print()
```

```python
# Cross-base top-k overlap.
#
# For each metric (prob, cos_sim, proj) and each (cond_a, cond_b) in
# {regular, intervened} x {regular, intervened}, report
#     | top_k(source, cond_a)  intersect  top_k(target, cond_b) |.
#
# The four combinations are reported in both source -> target and
# target -> source orderings; the values are symmetric (set intersection),
# but listing both makes the table easier to scan.

def _top_k_head_set(df, col, k=64):
    """Top-k unique (layer, head_num) pairs by `col` (descending)."""
    sorted_df = (
        df[['layer', 'head_num', col]]
        .sort_values(col, ascending=False)
        .drop_duplicates(subset=['layer', 'head_num'])
        .head(k)
    )
    return {(int(r['layer']), int(r['head_num'])) for _, r in sorted_df.iterrows()}


def compare_top_heads_across_bases(source_base, target_base, top_k=64, verbose=False):
    metric_sources = [
        ('prob',    prob_df, intervened_prob_df),
        ('cos_sim', sim_df,  intervened_sim_df),
        ('proj',    sim_df,  intervened_sim_df),
    ]

    for a_base, b_base in [(source_base, target_base), (target_base, source_base)]:
        print(f"=== {a_base} -> {b_base} (top {top_k}) ===")
        for suffix, reg_df, int_df in metric_sources:
            a_col = f"{a_base}_{suffix}"
            b_col = f"{b_base}_{suffix}"
            for a_cond, a_df in [("regular", reg_df), ("intervened", int_df)]:
                a_set = _top_k_head_set(a_df, a_col, k=top_k)
                for b_cond, b_df in [("regular", reg_df), ("intervened", int_df)]:
                    b_set    = _top_k_head_set(b_df, b_col, k=top_k)
                    shared   = a_set & b_set
                    n_shared = len(shared)
                    print(f"  [{suffix:<7}] {a_cond:<10} {a_base}  in  "
                          f"{b_cond:<10} {b_base}: {n_shared:>2}/{top_k}")
                    if verbose:
                        for layer, head in sorted(shared):
                            print(f"      (Layer {layer:>2}, Head {head:>2})")
        print()


compare_top_heads_across_bases('operand_1_1', 'operand_1_2', top_k=64)
```

```python
# For each (base, condition, metric):
#   - reduce each (layer, head_num) to a single value = its max across prompts
#     (matches the per-head ranking used in the cells above)
#   - report mean/median of that value over the top-64 heads and over ALL heads.

def head_value_summary(df, col, top_k=64):
    """Per-head representative value = max(col) across rows of that (layer, head_num).
    Returns mean/median over top_k heads and over all heads.
    """
    per_head = (
        df.groupby(['layer', 'head_num'])[col]
          .max()
          .sort_values(ascending=False)
    )
    top_vals = per_head.head(top_k)
    return {
        'top_mean':   top_vals.mean(),
        'top_median': top_vals.median(),
        'all_mean':   per_head.mean(),
        'all_median': per_head.median(),
        'n_heads':    len(per_head),
    }


bases = ['operand_1_1', 'operand_1_2', 'operand_2', 'factual_output']
metric_sources = [
    ('prob',    prob_df, intervened_prob_df),
    ('cos_sim', sim_df,  intervened_sim_df),
    ('proj',    sim_df,  intervened_sim_df),
]

TOP_K = 64
for base in bases:
    print(f"=== {base} ===")
    for suffix, reg_df, int_df in metric_sources:
        col = f"{base}_{suffix}"
        for cond_label, cond_df in [("regular", reg_df), ("intervened", int_df)]:
            s = head_value_summary(cond_df, col, top_k=TOP_K)
            print(
                f"  [{suffix:<7}] {cond_label:<10}: "
                f"top{TOP_K:<3} mean={s['top_mean']:+.4f} median={s['top_median']:+.4f}  |  "
                f"all (n={s['n_heads']:>4}) mean={s['all_mean']:+.4f} median={s['all_median']:+.4f}"
            )
    print()
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('experiments/attention_heads/output/GPT-OSS_stepwise/pattern_diff/h_pre_penultimate_sum.csv')

# Display basic information about the dataset
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst few rows:")
df.head()
```

```python
# Calculate average faithful_diff and unfaithful_diff for each layer
layer_summary = df.groupby('layer').agg({
    'faithful_diff': 'mean',
    'unfaithful_diff': 'mean'
}).round(6)

print("Average faithful_diff and unfaithful_diff by layer:")
print(layer_summary)

# Also show the count of observations per layer
layer_counts = df.groupby('layer').size()
print("\nNumber of observations per layer:")
print(layer_counts)
```

```python
# Collapse rows by layer and head_num, keeping only faithful_diff and unfaithful_diff averages
collapsed_df = df.groupby(['layer', 'head_num']).agg({
    'faithful_diff': 'mean',
    'unfaithful_diff': 'mean'
}).reset_index()

print("Collapsed dataset shape:", collapsed_df.shape)
print("\nFirst few rows of collapsed data:")
print(collapsed_df.head())

# Display the full collapsed dataset
collapsed_df
```

```python
# Find top 10 rows with highest faithful_diff
highest_faithful = collapsed_df.nlargest(10, 'faithful_diff')
print("Top 10 rows with highest faithful_diff:")
print(highest_faithful)

# Find top 10 rows with lowest faithful_diff
lowest_faithful = collapsed_df.nsmallest(10, 'faithful_diff')
print("\nTop 10 rows with lowest faithful_diff:")
print(lowest_faithful)

# Find top 10 rows with highest unfaithful_diff
highest_unfaithful = collapsed_df.nlargest(10, 'unfaithful_diff')
print("\nTop 10 rows with highest unfaithful_diff:")
print(highest_unfaithful)

# Find top 10 rows with lowest unfaithful_diff
lowest_unfaithful = collapsed_df.nsmallest(10, 'unfaithful_diff')
print("\nTop 10 rows with lowest unfaithful_diff:")
print(lowest_unfaithful)
```
